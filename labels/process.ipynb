```python
import numpy as np
import pandas as pd
```

```python
# Lấy danh sách file
from pathlib import Path

# Đường dẫn thư mục chứa ảnh
folder = Path("../dataset/raw")

# Các định dạng ảnh được hỗ trợ
extensions = {".jpg"}

# Lấy danh sách ảnh và sắp xếp theo tên hiện tại
images = sorted(
    [
        file.name for file in folder.iterdir()
        if file.is_file() and file.suffix.lower() in extensions
    ],
)
print(images)
```

```python
df = pd.DataFrame({'file_name': images, 'reading': ''})
print(df)
```

```python
# To CSV file
df.to_csv('annotation.csv', index=False)
```
