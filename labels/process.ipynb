```python
import numpy as np
import pandas as pd
```

```python
anno = pd.read_csv('annotation.csv')
```

```python
anno['filename'] = [f'sp_{i}.jpg'for i in range(1, 56)]
```

```python
anno.to_csv('annotation.csv')
```

```python
term_names = []
for i, name in enumerate(anno['filename']):
    term = str(name).replace('sp_', '')
    if len(term) == 5:
        term_names.append('sp_00'+term)
    elif len(term) == 6:
        term_names.append('sp_0'+term)
    elif len(term) == 7:
        term_names.append('sp_'+term)
print(term_names)
```

```python
anno['filename'] = term_names
```

```python
anno.to_csv('annotation.csv', index=False)
```
